```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import plotly.express as px
import plotly.graph_objects as go
import seaborn as sns
sns.set_style("whitegrid")
print("All libraries imported successfully!")
```

<h3>Loading Dataset:</h3>

```python
mql = pd.read_csv(
    "olist_marketing_qualified_leads_dataset.csv"
)
closed_deals = pd.read_csv(
    "olist_closed_deals_dataset.csv"
)
print("Both datasets loaded successfully!")
```

<h3>Checking Dataset Dimensions:</h3>

```python
print("Marketing Leads Shape:", mql.shape)
print("Closed Deals Shape:", closed_deals.shape)
```

<h3>Dataset Overview:</h3>

```python
display(mql.head())
display(closed_deals.head())
```

<h3>Checking Column Names:</h3>

```python
print("Marketing Leads Columns:")
print(mql.columns.tolist())

print("\nClosed Deals Columns:")
print(closed_deals.columns.tolist())
```

<h3>Checking Datatypes:</h3>

```python
print("Marketing Leads Information:")
mql.info()

print("\nClosed Deals Information:")
closed_deals.info()
```

<h3>Checking Missing Values:</h3>

```python
print("Missing Values in Marketing Leads:")
print(mql.isnull().sum())

print("\nMissing Values in Closed Deals:")
print(closed_deals.isnull().sum())
```

<h3>Checking Duplicate Records:</h3>

```python
print("Marketing Leads Duplicates:", mql.duplicated().sum())
print("Closed Deals Duplicates:", closed_deals.duplicated().sum())
```

<h3>Handling Missing Values in Marketing Lead:</h3>

```python
mql_clean = mql.copy()
closed_deals_clean = closed_deals.copy()
print("Dataset copies created successfully!")
```

<h3>Replacing Missing Values:</h3>

```python
mql_clean["origin"] = mql_clean["origin"].fillna("Unknown")
print(mql_clean.isnull().sum())
```

<h3>Handling Missing Values In Closed Deals:</h3>

```python
categorical_columns = [
    "business_segment",
    "lead_type",
    "lead_behaviour_profile",
    "has_company",
    "has_gtin",
    "average_stock",
    "business_type"
]
for column in categorical_columns:
    closed_deals_clean[column] = (
        closed_deals_clean[column].fillna("Unknown")
    )
print("Missing categorical values handled successfully!")
print(closed_deals_clean[categorical_columns].isnull().sum())
```

<h3>Converting Date Columns:</h3>

```python
mql_clean["first_contact_date"] = pd.to_datetime(
    mql_clean["first_contact_date"],
    errors="coerce"
)
closed_deals_clean["won_date"] = pd.to_datetime(
    closed_deals_clean["won_date"],
    errors="coerce"
)
print(mql_clean["first_contact_date"].dtype)
print(closed_deals_clean["won_date"].dtype)
```

<h3>Final data cleaning verification:</h3>

```python
print("Marketing Leads Shape:", mql_clean.shape)
print("Closed Deals Shape:", closed_deals_clean.shape)

print("\nMarketing Leads Duplicates:", mql_clean.duplicated().sum())
print("Closed Deals Duplicates:", closed_deals_clean.duplicated().sum())

print("\nMarketing Leads Missing Values:")
print(mql_clean.isnull().sum())

print("\nClosed Deals Missing Values:")
print(closed_deals_clean.isnull().sum())
```

<h3>Merging the Dataset:</h3>

```python
merged_data = pd.merge(
    mql_clean,
    closed_deals_clean,
    on="mql_id",
    how="left",
    indicator=True
)
display(merged_data.head())
print("Merged Dataset Shape:", merged_data.shape)
```

<h3>Creating Conversion Status:</h3>

```python
merged_data["conversion_status"] = np.where(
    merged_data["_merge"] == "both",
    "Converted",
    "Not Converted"
)
print(merged_data["conversion_status"].value_counts())
```

<h3>Calculating conversion and drop-off rates:</h3>

```python
total_leads = len(merged_data)
converted_leads = (
    merged_data["conversion_status"] == "Converted"
).sum()
not_converted_leads = total_leads - converted_leads
conversion_rate = (converted_leads / total_leads) * 100
dropoff_rate = (not_converted_leads / total_leads) * 100
print("Total Marketing Leads:", total_leads)
print("Converted Leads:", converted_leads)
print("Not Converted Leads:", not_converted_leads)
print("Conversion Rate:", round(conversion_rate, 2), "%")
print("Drop-off Rate:", round(dropoff_rate, 2), "%")
```

<h3>Analyzing marketing lead origins:</h3>

```python
origin_counts = mql_clean["origin"].value_counts().reset_index()
origin_counts.columns = ["origin", "total_leads"]
display(origin_counts)
```

<h3>Visualizing marketing lead origins:</h3>

<h3>Converted Leads by Origin:</h3>

```python
converted_data = merged_data[
    merged_data["conversion_status"] == "Converted"
]
converted_by_origin = (
    converted_data["origin"]
    .value_counts()
    .reset_index()
)
converted_by_origin.columns = [
    "origin",
    "converted_leads"
]
display(converted_by_origin)
```

```python
fig = px.bar(
    converted_by_origin,
    x="origin",
    y="converted_leads",
    title="Converted Leads by Marketing Origin",
    labels={
        "origin": "Marketing Origin",
        "converted_leads": "Converted Leads"
    },
    color="converted_leads",
    color_continuous_scale="Teal",
    text="converted_leads"
)
fig.update_layout(
    template="plotly_dark",
    title_x=0.5,
    xaxis_tickangle=-45
)
fig.update_traces(
    textposition="outside"
)
fig.show()
```

<h3>Calculating conversion rate by origin:</h3>

```python
total_by_origin = (
    merged_data.groupby("origin")["mql_id"]
    .count()
)
converted_by_origin_count = (
    merged_data[
        merged_data["conversion_status"] == "Converted"
    ]
    .groupby("origin")["mql_id"]
    .count()
)
origin_analysis = pd.DataFrame({
    "Total Leads": total_by_origin,
    "Converted Leads": converted_by_origin_count
})
origin_analysis["Converted Leads"] = (
    origin_analysis["Converted Leads"].fillna(0)
)
origin_analysis["Conversion Rate (%)"] = (
    origin_analysis["Converted Leads"]
    / origin_analysis["Total Leads"]
) * 100
origin_analysis = origin_analysis.sort_values(
    by="Conversion Rate (%)",
    ascending=False
)
display(origin_analysis.round(2))
```

```python
origin_plot = origin_analysis.reset_index()
fig = px.bar(
    origin_plot,
    x="origin",
    y="Conversion Rate (%)",
    title="Conversion Rate by Marketing Origin",
    labels={
        "origin": "Marketing Origin",
        "Conversion Rate (%)": "Conversion Rate"
    },
    color="Conversion Rate (%)",
    color_continuous_scale="Viridis",
    text=origin_plot["Conversion Rate (%)"].round(2).astype(str) + "%"
)
fig.update_layout(
    template="plotly_dark",
    title_x=0.5,
    xaxis_tickangle=-45
)
fig.update_traces(
    textposition="outside"
)
fig.show()
```

<h3>Marketing origin performance:</h3>

```python
merged_data["origin"] = (
    merged_data["origin"]
    .astype(str)
    .str.strip()
    .str.lower()
)
origin_performance = (
    merged_data.groupby("origin")
    .agg(
        Total_Leads=("mql_id", "count"),
        Converted_Leads=(
            "conversion_status",
            lambda x: (x == "Converted").sum()
        )
    )
    .reset_index()
)
origin_performance["Conversion_Rate (%)"] = (
    origin_performance["Converted_Leads"]
    / origin_performance["Total_Leads"]
) * 100
origin_performance["Drop_off_Rate (%)"] = (
    100 - origin_performance["Conversion_Rate (%)"]
)
origin_performance = origin_performance.sort_values(
    "Conversion_Rate (%)",
    ascending=False
)
display(origin_performance.round(2))
```
